# Reverse a string
### This coding challenge is called reverse_string

```python
def reverse_string(first_string):
    reversed_string = first_string[::-1]
    return reversed_string
```

# Testing reverse string function

```python
reverse_string("chrissy")
```

# Palindrome
## This coding challenge checks whether a given string is a palindrome
### We will use the reverse string function from above

```python
def is_pallindrome(user_input):
    user_input = user_input.lower()
    user_input = ''.join(char for char in user_input if char.isalnum())
    reverse_input = reverse_string(user_input)
    if user_input == reverse_input:
        print("True")
    else:
        print("False")
```

# Testing pallindrome function

```python
is_pallindrome("racecar")
```

# Find Max
## This coding challenge takes a list of numbers and returns the maximum value in the list

```python
def find_max(numbers):
    try:
        if not numbers:
            raise ValueError("The list is empty")

        max_value = numbers[0]

        for num in numbers[1:]:
            if num > max_value:
                max_value = num

        return max_value

    except ValueError:

        print("Please provide a list that is not empty.")
        
        return None
```

# Testing find_max

```python
find_max([10.2,16,4,3,20.5,5,7,8])
```

```python
find_max([])
```

# Is Prime
## This function checks if a number is prime

```python
def is_prime(n):
    if n < 2:
        return False
    for number in range(2, n):
        if n % number == 0:
            return False
    return True
```

# Testing is_prime

```python
is_prime(20)    
```

# Caesar Cipher
## Returns encoded message

```python
def caesar_cipher(input,shift_value):

    input = str(input) 
    cipher=""

    for character in input:
        if 'a' <= character <='z':
            cipher_order = (ord(character) - ord('a') + int(shift_value)) % 26 + ord('a')
            cipher += chr(cipher_order)

        elif 'A' <= character <= 'Z':
            cipher_order = (ord(character) - ord('A') + int(shift_value)) % 26 + ord('A')
            cipher += chr(cipher_order)

        elif '0' <= character <= '9':
            cipher_order = (ord(character) - ord('0') + shift_value) % 10 + ord('0')
            cipher += chr(cipher_order)
            
        else:
            cipher += character
        
    return cipher
```

# Testing Caesar Cipher

```python
caesar_cipher('Chrissy', 2)
```
